```python
import pandas as pd
file = "/home/ubuntu/exp/services/httpmutator-benchmark/expScripts/mutationInfo/GenomeNexus/post-annotation.csv"
# read csv with no header, specify headers
df = pd.read_csv(
    file,
    header=None,                  # 文件无表头
    names=[
        "filename",
        "className",
        "mutator",
        "method",
        "lineNumber",
        "status",
        "description",
        "killingTest"
    ],
    dtype={
        "status": "category"
    }
)
```

```python
NO_COV = "NO_COVERAGE"

all_no_cov_by_class = (
    df
    .groupby("className")["status"]
    .apply(lambda s: (s == NO_COV).all())
)
classes_all_no_cov = all_no_cov_by_class[all_no_cov_by_class].index.tolist()

# print("Classes with all mutations uncovered:")
print('''<excludedClasses>''')
for cls in classes_all_no_cov:
    print(f"         <excludedClass>{cls}</excludedClass>")
print('''</excludedClasses>''')
```

```python
# now, creat a new dataframe without these no-coverage class
filtered_df = df[~df["className"].isin(classes_all_no_cov)].copy()
```

```python
NO_COV = "NO_COVERAGE"

methods_all_no_cov = (
    filtered_df
    .groupby("method")["status"]
    .apply(lambda s: (s == NO_COV).all())
    .loc[lambda x: x]
    .index
    .tolist()
)


print('''<excludedMethods>''')
for m in methods_all_no_cov:
    print(f"         <excludedMethod>{m}</excludedMethod>")
print('''</excludedMethods>''')
```
